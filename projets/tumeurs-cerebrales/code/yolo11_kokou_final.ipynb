## 2e entraînement

```python
yaml_path = '/content/brain-tumor-mri-2/data.yaml'

# Sauvegarder le fichier YAML en local
custom_yaml_path = '/content/custom_yolov11.yaml'
with open(custom_yaml_path, 'w') as f:
    f.write(custom_yaml_content)

# Initialiser le modèle avec la nouvelle architecture et charger les poids de base
model = YOLO(custom_yaml_path).load('yolo11s.pt')

results = model.train(
    data=yaml_path,

    epochs=50,
    imgsz=640,
    batch=16,

    optimizer='AdamW',

    lr0=0.0005,        # 🔥 plus stable
    lrf=0.01,

    weight_decay=0.0005,

    warmup_epochs=5,

    mosaic=1.0,
    mixup=0.1,

    box=5.0,           # 🔥 réduit
    dfl=1.5,           # 🔥 réduit

    project='/content/drive/MyDrive/BrainTumor',
    name='YOLOv11_clean_box_tuned',
)
```

## fine-tuning

```python
model = YOLO('/content/drive/MyDrive/BrainTumor/YOLOv11_clean_box_tuned-2/weights/best.pt')

model.train(
    data=yaml_path,
    epochs=20,

    lr0=0.0001,     # 🔥 plus petit
    lrf=0.01,

    mosaic=0.0,     # 🔥 IMPORTANT
    mixup=0.0,
    copy_paste=0.0,

    box=8.0,
    dfl=2.0,

    project='/content/drive/MyDrive/BrainTumor',
    name='YOLOv11_finetuned',
)
```

## Evaluation du modèle

```python
metrics = model.val()
```
